```python
# federal_workers_cd_selections.csv 전처리 코드

import pandas as pd

# 파일 불러오기
file_path = 'G:/내 드라이브/2025-1/경제학연습//federal_workers_cd_selections.csv'
df = pd.read_csv(file_path)

# 1. 필요한 컬럼만 선택 (state_name, federal_workers, share_of_employment)
#    - federal_workers는 주별 총합 필요
#    - share_of_employment는 주별 평균 또는 가중평균 필요(선택사항)

# 2. 주별로 federal_workers 합계 계산
state_federal_workers = df.groupby('state_name')['federal_workers'].sum().reset_index()

# 3. share_of_employment 처리
# (a) 가장 간단하게: 주별 district들의 share_of_employment 평균 내기
state_share_of_employment = df.groupby('state_name')['share_of_employment'].mean().reset_index()

# 4. 합친다
processed_df = pd.merge(state_federal_workers, state_share_of_employment, on='state_name')

# 5. 정렬 (선택사항)
processed_df = processed_df.sort_values('state_name').reset_index(drop=True)

# 6. 결과 확인
print(processed_df.all)

# 7. 저장 (선택사항)
# processed_df.to_csv('processed_federal_workers_by_state.csv', index=False)
```

```python
# federal_workers_cd_selections.csv 전처리 및 DOGE intensity 계산 코드

import pandas as pd

# 파일 불러오기
file_path = 'G:/내 드라이브/2025-1/경제학연습/federal_workers_cd_selections.csv'
df = pd.read_csv(file_path)

# 1. 필요한 컬럼만 선택 (state_name, federal_workers, share_of_employment)

# 2. 주별로 federal_workers 합계 계산
state_federal_workers = df.groupby('state_name')['federal_workers'].sum().reset_index()

# 3. share_of_employment 처리
# (a) 가장 간단하게: 주별 district들의 share_of_employment 평균 내기
state_share_of_employment = df.groupby('state_name')['share_of_employment'].mean().reset_index()

# 4. 합친다
processed_df = pd.merge(state_federal_workers, state_share_of_employment, on='state_name')

# 5. DOGE intensity 계산
federal_employment_cut_rate = 0.12  # 연방 고용 감축률 12%
federal_budget_cut_rate = 0.026     # 연방 예산 감축률 2.6%

processed_df['DOGE_Intensity_EmploymentBased'] = processed_df['share_of_employment'] * federal_employment_cut_rate
processed_df['DOGE_Intensity_BudgetBased'] = processed_df['share_of_employment'] * federal_budget_cut_rate

# 6. 정렬
processed_df = processed_df.sort_values('state_name').reset_index(drop=True)

# 7. 결과 확인
print(processed_df[['state_name', 'DOGE_Intensity_EmploymentBased', 'DOGE_Intensity_BudgetBased']])

# 8. 저장 (선택사항)
processed_df.to_csv('processed_federal_workers_with_doge_intensity.csv', index=False)
```

```python
processed_df['DOGE_Intensity_WeightedAverage'] = (
    0.5 * processed_df['DOGE_Intensity_EmploymentBased'] +
    0.5 * processed_df['DOGE_Intensity_BudgetBased']
)

# 정렬 (선택사항)
processed_df = processed_df.sort_values('state_name').reset_index(drop=True)

# 결과 출력
print(processed_df[['state_name', 'DOGE_Intensity_WeightedAverage']])

# 필요한 경우 저장
processed_df.to_csv('doge_intensity_weighted_average_full.csv', index=False)
```
